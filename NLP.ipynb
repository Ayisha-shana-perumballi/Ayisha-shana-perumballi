```python
# Install required libraries (Run once)
# !pip install nltk

import nltk
import string

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

text = """
My name is Ayisha Shana Perumballi. I am from Malappuram, India. I completed my schooling at JSR hss.
I belong to a supportive family that has always encouraged me to learn and grow.
I am currently a final-year B.Tech student specializing in Artificial Intelligence and Data Science at
Adi Shankara Institute of Engineering and Technology.
I have participated in several hackathons, coordinated technical and cultural events,
and worked on AI-based projects.
My hobbies include designing presentations, exploring AI tools, listening to music,
and learning new technologies.
I am interested in Artificial Intelligence, Data Analytics, and Machine Learning.
My strengths are creativity, teamwork, quick learning, and problem-solving.
My weaknesses are that I sometimes overthink and tend to spend extra time perfecting my work.
"""

print("ORIGINAL TEXT\n")
print(text)


lower_text = text.lower()
print("\n\n1. LOWERCASE\n")
print(lower_text)


sentences = sent_tokenize(lower_text)
print("\n\n2. SENTENCE TOKENIZATION\n")
print(sentences)


words = word_tokenize(lower_text)
print("\n\n3. WORD TOKENIZATION\n")
print(words)


words_no_punct = [word for word in words if word not in string.punctuation]
print("\n\n4. AFTER REMOVING PUNCTUATION\n")
print(words_no_punct)


stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words_no_punct if word not in stop_words]
print("\n\n5. AFTER REMOVING STOPWORDS\n")
print(filtered_words)


stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print("\n\n6. STEMMING\n")
print(stemmed_words)


lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
print("\n\n7. LEMMATIZATION\n")
print(lemmatized_words)
```
